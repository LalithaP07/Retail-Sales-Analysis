Hypothesis 1 : Which product categories in the store have the biggest profit margins?

```python
#Display Column labels
data.columns
```

```python
# Grouping by category and calculating the profit for each category
cat_profit=data.groupby('Category')['Profit'].sum()
cat_profit
```

```python
cat_profit=data.groupby('Category')['Profit'].sum()
#plot bar chart to visualize total profit for each category
cat_profit.plot(kind='bar')
# Add titles and labels to the plot
plt.title("profit by Category")
plt.xlabel("Category")
plt.ylabel("total profit")
#Display the plot
plt.show()

#Conclusion : The technology products have the highest profit margin of the three categories.
```

Hypothesis 3 : How do sales fluctuate by product category throughout the course of the year?

```python
data.columns
```

```python
#Create new column 'Order Month' based on the month of 'Order Date'
data['Order Month']=pd.DatetimeIndex(data['Order Date']).month
month_sales=data.groupby('Order Month')['Sales'].sum()
#Plotting line chart
month_sales.plot(kind='line')
# Add titles and labels to the plot
plt.title("Total Sales by Month")
plt.xlabel("Month")
plt.ylabel("Total Sales")
#Display the plot
plt.show()

#Conclusion : The sales are higher during certain months of the year
```

Hypothesis 4 : what is the rate of returned products for orders with same-day shopping compared to other shopping options?

```python
data.columns
```

```python
#Calculate total number of orders for each shipping mode
total_orders_by_shipping_mode=data.groupby('Ship Mode').size()
#Filtering data that include returned orders
returned_orders_by_shipping_mode=data[data['Profit']<0].groupby('Ship Mode').size()
#Calculate the return percentage for each shipping mode
returned_per_by_shipping_mode=(returned_orders_by_shipping_mode/total_orders_by_shipping_mode)*100
#Print return percentage for each shipping
print(returned_per_by_shipping_mode)
#Plot bar chart
returned_per_by_shipping_mode.plot(kind="bar")
# Add titles and labels to the plot
plt.title("Return Per By Shipping Mode")
plt.xlabel("Shipping Mode")
plt.ylabel("Return Per")
#Display the plot
plt.show()

#Conclusion : The orders with the same-day shipping have the lowest rate of returned products
```

Hypothesis 5 : How do weekday sales and profits compare to weekend sales by product category?

```python
data.columns
```

```python
#Create column in 'Oreder Day' based on the day of the week
data['Order Day']=pd.DatetimeIndex(data['Order Date']).day_name()
#Grouping data by 'Order Day' column and calculate sum of profit for each day of week
day_profit=data.groupby('Order Day')['Profit'].sum()
#Plot bar chart
day_profit.plot(kind="bar")
# Add titles and labels to the plot
plt.title("Total Profit by the day of the week")
plt.xlabel("Day of the week")
plt.ylabel("Total Profit")
#Display the plot
plt.show()

#Conclusion : Company's profit is higher on weekdays than compared to weekends.
```
